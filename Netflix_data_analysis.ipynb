```python
# 1. IMPORT LIBRARIES
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. LOAD DATASET
df = pd.read_csv("netflix_titles.csv")

print("Dataset Shape:", df.shape)
print(df.head())

# 3. DATA CLEANING
# Check missing values
print("\nMissing Values:\n", df.isnull().sum())

# Fill missing values
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')

# Drop rows with critical missing values
df.dropna(subset=['date_added', 'rating'], inplace=True)

# Convert date column
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce', format='mixed')

# Create new columns
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month

# 4. EXPLORATORY DATA ANALYSIS

# Movies vs TV Shows
type_counts = df['type'].value_counts()
print("\nContent Type:\n", type_counts)

# Top 10 Genres
genres = df['listed_in'].str.split(', ')
genre_list = genres.explode()
print("\nTop Genres:\n", genre_list.value_counts().head(10))

# Content added per year
yearly_content = df['year_added'].value_counts().sort_index()
print("\nContent added per year:\n", yearly_content)

# Top countries
print("\nTop Countries:\n", df['country'].value_counts().head(10))

# 5. VISUALIZATIONS

# 1. Movies vs TV Shows
plt.figure()
type_counts.plot(kind='bar')
plt.title("Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()

# 2. Top Genres
plt.figure()
genre_list.value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Genres")
plt.show()

# 3. Content Trend Over Years
plt.figure()
yearly_content.plot(kind='line', marker='o')
plt.title("Content Added Over Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.grid(True)
plt.show()

# 4. Top Countries
plt.figure()
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Content Producing Countries")
plt.show()

# 6. ADVANCED ANALYSIS

# Longest Movies
movies = df[df['type'] == 'Movie']
movies['duration'] = movies['duration'].str.replace(' min', '').astype(float)

print("\nTop 5 Longest Movies:\n",
      movies.sort_values(by='duration', ascending=False)[['title', 'duration']].head())

# TV Shows with most seasons
shows = df[df['type'] == 'TV Show']
shows['seasons'] = shows['duration'].str.replace(' Seasons', '').str.replace(' Season', '').astype(float)

print("\nTop 5 TV Shows with Most Seasons:\n",
      shows.sort_values(by='seasons', ascending=False)[['title', 'seasons']].head())

# 7. SIMPLE RECOMMENDATION SYSTEM

def recommend_by_genre(genre):
    result = df[df['listed_in'].str.contains(genre, case=False, na=False)]
    return result[['title', 'listed_in']].head(10)

print("\nRecommendations (Comedy):\n", recommend_by_genre("Comedy"))


# 8. BUSINESS INSIGHTS

print("\n--- BUSINESS INSIGHTS ---")
print("1. Most content type:", type_counts.idxmax())
print("2. Most popular genre:", genre_list.value_counts().idxmax())
print("3. Peak content year:", yearly_content.idxmax())
print("4. Top country:", df['country'].value_counts().idxmax())
```
